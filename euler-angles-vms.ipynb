# Introduction
In the previous section, we presented an analytical expression for the nutation angle $\theta$ in Equation {eq}`eq39` and repeated here for convenience:
```{math}
:label: eq_euler_angles_theta_vms
\theta = \tan^{-1}\bigg(\frac{I \omega_{12}}{J \omega_3} \bigg) = \tan^{-1}\bigg(\frac{\Gamma k_1^2 \omega_{0}}{k_3^2 \omega_3} \bigg) 
```
This is useful in evaluating the pointing stability of a spacecraft- this has long been possible for 
axisymmetric constant mass sytems as the conditions for conservation of angular momentum are
straightforward. Recently, a proof was developed to show that the angular momentumvector of
variable mass systems ${\bf H}^*$ is inertially fixed despite not being a conserved quantity.
Its utility in studying the pointing stability of axisymmetric variable mass systems was
presented in {cite}`nanjangud2018geometry`.

The angular momentum conservation principle is also powerful in geometrically deriving analytrical
solutions to the Euler angles for axisymmetric systems; this is well documented for constant mass
systems but is yet to be explored for variable mass systems. 

# Euler Angle Solutions
To derive analytical solutions to the Euler angles, we introduce two more inertially fixed unit vectors ${\bf n}_1$ and ${\bf n}_2$ that are orthogonal to ${\bf n}_3$ that make up the inertial fixed frame.
Then, a $3-1-3$ rotation sequence is used to reorient the body-fixed frame with the inertial frame 
by an amount of $\psi$, $\theta$, and $\phi$, respectively. This $\theta$ is the same as the 
nutation angle derived in the previous section; which is why we say that it is also the
solution to the second Euler angle.

Next, we will derive analytical expressions for the first and third Euler angles for variable mass 
systems ($\psi$ and $\phi$, respectively) which has not previously been addressed in the literature.

````python
from sympy.physics.mechanics import dynamicsymbols
from sympy.physics.vector import init_vprinting
init_vprinting(use_latex='mathjax', pretty_print=False)
from sympy import atan, symbols
I, J, omega_12, omega_3 = dynamicsymbols('I J omega_12 omega_3')
theta = atan(I*omega_12/J*omega_3)
theta_dot = theta.diff().simplify()
# theta_dot
````

To do so, we will first extract the angular velocity vector in terms of Euler angle rates as:
```{math}
:label: eq_euler_angles_omega_vms
\boldsymbol\omega &= \dot{\psi} {\bf k}_3 + \dot{\theta} {\bf k}_1 + \dot{\phi} {\bf j}_{3}\\
&= \dot{\theta} {\bf k}_1 + \dot{\phi} \sin \theta {\bf k}_{2} + (\dot{\psi} + \dot{\phi} \cos \theta) {\bf k}_3 
```

Equating $\boldsymbol\omega$ given in the RHS of Equation {eq}`eq_euler_angles_omega_vms` to the RHS of Equation {eq}`eq32` gives
```{math}
:label: eq_euler_angles_omega_vms_3
\dot{\theta} {\bf k}_1 + \dot{\phi} \sin \theta {\bf k}_{2} + (\dot{\psi} + \dot{\phi} \cos \theta) {\bf k}_3 = \omega_0 \Gamma{\bf b}_{12} + \omega_3 {\bf b}_3,
```
but for the assumed Euler 3-1-3 rotation sequence, we have $\hat{\bf b}_3 = {\bf k}_3$ thus
```{math}
\omega_3 = \dot\psi + \dot{\phi} \cos \theta.
```
Consequently, we have
```{math}
\dot{\theta} {\bf k}_1 + \dot{\phi} \sin \theta {\bf k}_{2} = \omega_0 \Gamma{\bf b}_{12} 
```
which can be expressed in terms of the body-fixed frame as
```{math}
\dot{\theta} (-\sin\psi {\bf b}_1 + \cos\psi {\bf b}_2)  + \dot{\phi} \sin \theta (\cos\psi {\bf b}_1 + \sin\psi {\bf b}_2) = \omega_0 \Gamma {\bf b}_{12}.
```
If $\psi$ is arbitraily chosen to be identical to $\chi$, then 
```{math}
\dot{\theta} (-\sin\chi {\bf b}_1 + \cos\chi {\bf b}_2)  + \dot{\phi} \sin \theta (\cos\chi {\bf b}_1 + \sin\chi {\bf b}_2) = \omega_0 \Gamma {\bf b}_{12}
```
or
```{math}
\dot{\theta} {\bf b}_t  + \dot{\phi} \sin \theta {\bf b}_{12} = \omega_0 \Gamma {\bf b}_{12}.
```
which then gives us:

```{math}
\dot{\phi} \sin \theta = \omega_0 \Gamma.
```


Now, we can make the assumption that ${\bf k}_2$ is aligned with ${\bf b}_{12}$ and
therefore ${\bf k}_1$ is aligned with a unit vector orthogonal to both ${\bf b}_{12}$ and ${\bf b}_3$, which will allow us to solve for $\dot{\psi}$ and $\dot{\phi}$.


If now we make the special choice of the arbitrary inertially fixed unit 
vector ${\bf n}_{3}$ so as to align it with the angular momentum
${\bf H}^*$, then we see that the nutation angle $\theta$ is also the
second Euler angle (i.e., the angle between ${\bf b}_{3}$ and
${\bf H}^*$ in {numref}`fig-4-3`).

If axisymmetry is assumed for both constant mass and variable mass systems, then $\theta$ is identical
in form to {eq}`eq39`; however, $\dot{\theta} = 0$ only for the former
given that $I, J$ and $\omega_3$ are constant. For a general axisymmetric variable mass system,
$\dot{\theta} \neq 0$ and the expression for $\tan \theta$ is dependent on burn geometries but
a general expression for $\dot{\theta}$ is given by:
```{math}
:label: eq_euler_angles_theta_dot
\dot{\theta} = \frac{\left(I \omega_{12} \dot{\omega}_3 + I \omega_3 \dot{\omega}_{12} + \omega_{12} \omega_3 \dot{I}\right) J - I \omega_{12} \omega_3 \dot{J}}{I^2 \omega_{12}^2 \omega_3^2 + J^2}
```

Discussions on three specific burns will be presented in the following sections but we proceed to
derive analytical forms for the second and third Euler angle rates, $\dot{\psi}$ and $\dot{\phi}$
so that we can then solve for the first and third Euler angles, $\psi$ and $\phi$, analytically
for each of the three burns.

## Deriving $\dot{\phi}$ and $\dot{\psi}$ from $\boldsymbol\omega$

Equating $\boldsymbol\omega$ given in Equation {eq}`eq_euler_angles_omega_vms` to that in Equation {eq}`eq32` gives us
```{math}
:label: eq_euler_angles_omega_vms_3
\dot{\theta} {\bf k}_1 + \dot{\phi} \sin \theta {\bf k}_{2} + (\dot{\psi} + \dot{\phi} \cos \theta) {\bf k}_3 = \omega_0 \Gamma{\bf b}_{12} + \omega_3 {\bf b}_3,
```

but noting that $\hat{\bf b}_3 = {\bf k}_3$, we have
```{math}
\omega_3 = \dot\psi + \dot{\phi} \cos \theta
```
which tells us that 

```{math}
:label: eq_euler_angles_dot_psi
\dot{\psi} = \omega_3 - \dot{\phi} \cos \theta.
```

From Equation {eq}`eq_euler_angles_omega_vms`, we also have
```{math}
:label: eq_euler_angles_dot_theta_b12
\dot{\theta} {\bf k}_1 + \dot{\phi} \sin \theta {\bf k}_{2} = \omega_0 \Gamma{\bf b}_{12} 
```

or

```{math}
:label: eq_euler_angles_dot_theta_b12_vms
\dot{\theta} {\bf k}_1 + \dot{\phi} \sin \theta {\bf k}_{2} = \omega_0 \Gamma \cos\chi {\bf b}_{1} + \omega_0 \Gamma \cos\chi {\bf b}_{2}
```

After expressing the vector on the LHS Equation {eq}`eq_euler_angles_dot_theta_b12_vms`
in terms of the body-fixed frame, we get the following two scalar equations:
```{math}   
:label: eq_euler_angle_rates_b1
\sin(\psi) \sin(\theta) \dot{\phi} + \cos(\psi) \dot{\theta} = \omega_0 \Gamma \cos\chi
```
along the ${\bf b}_1$ direction and

```{math}
:label: eq_euler_angle_rates_b2
\cos(\psi) \sin(\theta) \dot{\phi} - \sin(\psi) \dot{\theta} = \omega_0 \Gamma \sin\chi
```
along the ${\bf b}_2$ direction.

Noting that $\dot{\theta}$ is given by Equation {eq}`eq_euler_angles_theta_dot`, the two Equations {eq}`eq_euler_angle_rates_b1`
and {eq}`eq_euler_angle_rates_b2` are linear in $\dot{\phi}$ and $\dot{\theta}$ and can be easily simultaneously for
$\dot{\phi}$ to yield:
```{math}
:label: eq_euler_angle_rates_phi
\dot{\phi} \sin\theta= \omega_0 \Gamma \bigl(\cos\chi \sin\psi - \sin\chi \cos\psi \bigr).
```


```{math}
:label: eq_euler_angle_rates_phi
\dot{\phi} = \omega_0 \Gamma \frac{\bigl(\cos\chi \sin\psi - \sin\chi \cos\psi \bigr)}{\sin\theta}.
```

And for completeness, we rewrite the expression for $\dot{\psi}$ given above 
in Equation {eq}`eq_euler_angles_dot_psi` as:
```{math}
\dot{\psi} = \omega_3 - \dot{\phi} \cos \theta
```
where we can substitute the expression for $\dot{\phi}$ from Equation {eq}`eq_euler_angle_rates_phi` to get:
```{math}
\dot{\psi} = \omega_3 - \omega_0 \Gamma \bigl(\cos\chi \sin\psi - \sin\chi \cos\psi \bigr) \cos \theta.
```

We can now solve for the first Euler angle, $\phi$, as the following integral:
```{math}
:label: eq_euler_angles_phi_vms
\phi = \int \dot{\phi} dt = \int \omega_0 \Gamma \bigl(\cos\chi \sin\psi - \sin\chi \cos\psi \bigr) dt.
```

````python
# Extract inertia change equations for a uniform burn cylinder
def inertia_changes_uniform_burn(m, m_dot, R, h, z_e):
    """
    Calculate the time derivatives of mass, moments of inertia, and geometry for a uniformly burning cylinder.
    
    Parameters:
    m (float): Current mass of the cylinder
    m_dot (float): Mass flow rate (negative for mass loss)
    R (float): Radius of the cylinder
    h (float): Half-length of the cylinder
    z_e (float): Distance from the center of mass to the nozzle exit plane
    
    Returns:
    tuple: (md, Id, Jd, z_e_dot, R_dot) where
        md (float): Time derivative of mass
        Id (float): Time derivative of transverse moment of inertia
        Jd (float): Time derivative of spin moment of inertia
        z_e_dot (float): Time derivative of z_e (always 0 for uniform burn)
        R_dot (float): Time derivative of R (always 0 for uniform burn)
    """
    md = m_dot  # mass flow rate
    Id = md * (R**2 / 4 + h**2 / 3)  # transverse moment of inertia derivative
    Jd = md * R**2 / 2  # spin moment of inertia derivative
    
    # For a uniform burn cylinder, z_e and R remain constant
    z_e_dot = 0
    R_dot = 0
    
    return md, Id, Jd, z_e_dot, R_dot

# Example usage:
# md, Id, Jd, z_e_dot, R_dot = inertia_changes_uniform_burn(m, m_dot, R, h, z_e)

# Note: In a uniform burn, the cylinder's dimensions (R and h) remain constant,
# and the center of mass position relative to the nozzle exit (z_e) also remains constant.
# This is why z_e_dot and R_dot are always 0 for this type of burn.





````



Evaluation of these angles $\theta$ and $\beta$ along with the solutions to the angular velocities permit a visual understanding of the motion of the body in both inertial space and the body-fixed coordinate system.

In the case of axisymmetric rigid bodies of constant mass, $\Gamma = 1$ and the spin rate and inertia scalars are constant. Consequently, in Equation {eq}`eq39` and {eq}`eq40`, both angles evaluate to constants and visualizing the motion of $\boldsymbol{\omega}$ results in the formation of two cones: as it rotates about ${\bf b}_3$ and another as it rotates about ${\bf n}_h$, a unit vector parallel to the inertially fixed ${\bf H}^*$. The two cones are appropriately named the _body cone_ and _space cone_, respectively.

In the case of axisymmetric variable mass systems, angles $\theta$ and $\beta$ are typically functions of time. The angles may grow or decay as dictated by the instantaneous values of the radii of gyration, $\Gamma$, and the spin rate. It is also possible, in the case of the same variable mass system, for $\beta$ to exceed $\theta$ at certain instants while it may not at certain other instants. {numref}`fig-4-3` represents an instant where $\beta>\theta$, which occurs when $k_3>k_1$. Also, $\boldsymbol{\omega}$ does not trace cones as it rotates about ${\bf b}_3$ and ${\bf n}_h$. Thus, generic names are used for the surfaces it traces; in this document, they are referred to as the _body surface_ and _space surface_ (which reduce to the _body cone_ and _space cone_ for the limiting case of a constant mass rigid body, respectively).

In the spacecraft dynamics literature, $\theta$ is referred to as the nutation angle {cite}`kaplan`. In classical mechanics, $\theta$ may also be identified as the second Euler angle {cite}`likinstext` and, thus, Equation {eq}`eq39` also gives the solution to said Euler angle. Growth in nutation angles results in a coning motion which is undesirable in spacecraft. Potential hazards of excessive nutation or coning motion might be poor initial conditions for subsequent mission operations, or possible loss of communications with a spacecraft due to incorrect pointing between its antenna and the base station. One such case of excessive nutation growth was observed in missions powered by the STAR-48 SRM {cite}`meyer, flandro, cochran, or, mingorI, mingori2`. To mitigate this growth, an active nutation damper is used in these missions {cite}`webster`.

Closed form expressions to $\theta$ are available when the angular speeds are themselves available in closed form. These closed-form expressions are invaluable to not only evaluate stability but also for appropriate control strategy formulation to temper any instabilities. Solutions to the angular speeds are derived in the next section for a special class of variable mass systems: the axisymmetric variable mass cylinder. These closed form solutions are then utilized in constructing the body and space surfaces. The choice of cylindrical systems as an idealization is based on their use in several prior studies on rocket flight; Snyder and Warner {cite}`snyder` and Eke et al. {cite}`ekewang2,ekemao2` are some examples of investigators who have incorporated cylindrical propellant grains in their studies on the attitude motions of variable mass space vehicles.
